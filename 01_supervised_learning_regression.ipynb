# Supervised Learning

## What is Supervised Learning?

Supervised learning means:

> We give the machine **input data (X)** and the **correct answers (y)**, and the machine learns the relationship between them.

Let's understand this with a simple example.

### Example: House Prices

| Area | Bedrooms |      Price |
| ---: | -------: | ---------: |
| 1000 |        2 |  5,000,000 |
| 1500 |        3 |  7,000,000 |
| 2000 |        4 | 10,000,000 |

Here:

* **X = information about the house** (features)
* **y = correct price** (target)

So the basic idea is:

**X → Model → y**

The model learns from examples where we already know the correct answer.

---

## Why is it called "Supervised"?

It is called **supervised learning** because the correct answers (labels/targets) are available during training.

For example:

| Input      | Correct Answer |
| ---------- | -------------: |
| 1000 sq ft |      5 million |
| 1500 sq ft |      7 million |
| 2000 sq ft |     10 million |

The model looks at these examples and tries to learn the pattern between the input and the correct answer.

---

## Making a Prediction

After learning from the previous examples, we give the model a **new house**:

**1800 sq ft**

The model might predict:

**Predicted price → 8.5 million**

The model has never seen this exact house before. It uses the patterns it learned from the previous examples to make a prediction.

# What is Regression?

Regression is a type of **supervised machine learning** where we predict a **numerical value** based on input features.

### Example

Suppose we want to predict a student's marks based on the number of hours they studied.

* **x = 2 hours studied**
* **y = 55 marks** → actual output/target

After training, if we give the model a new input:

* **x = 5 hours studied**

The model might predict:

* **ŷ = 78 marks** → predicted value

Here:

* **y** = actual value
* **ŷ (y-hat)** = predicted value

The goal of regression is to make the **predicted value (ŷ)** as close as possible to the **actual value (y)**.

* `x` = input feature
* `w` = weight/slope
* `b` = bias/intercept
* `ŷ` = predicted value

### Think of it as a prediction machine

We give the model an input `x`.

The model applies:

**ŷ = wx + b**

and produces a predicted value `ŷ`.

```text
Input x

wx + b

Predicted value ŷ
```


**ŷ = 52 marks**

**y = 55 marks**

Therefore:

```text
Actual value     y  = 55
Predicted value  ŷ  = 52
```

`y` and `ŷ` are **not the same thing**.

* `y` → the actual value from our dataset
* `ŷ` → the value predicted by our model

# Cost Function

Now we have a prediction from our model, but we need a way to answer an important question:

> **How good or bad are our predictions?**

The **Cost Function** helps us measure how far our predictions are from the actual values.

## Cost Function Formula

The cost function for linear regression is:

$$
J(w,b)=\frac{1}{2m}\sum_{i=1}^{m}\left(\hat{y}^{(i)}-y^{(i)}\right)^2
$$

Where:

* $J(w,b)$ = cost of the model
* $m$ = number of training examples
* $\hat{y}^{(i)}$ = predicted value for the $i$-th training example
* $y^{(i)}$ = actual value for the $i$-th training example
* $\hat{y}^{(i)}-y^{(i)}$ = difference between the prediction and actual value
* $\left(\hat{y}^{(i)}-y^{(i)}\right)^2$ = squared error

---

## What is happening in the Cost Function?

Suppose we have **5 training examples** and therefore 5 predicted values.

For each example, we first calculate the difference between the **predicted value** and the **actual value**:

$$
\hat{y}^{(i)}-y^{(i)}
$$

For example:

```text
Actual y       = 55
Predicted ŷ    = 52

Error = 52 - 55
      = -3
```

### Why do we square the error?

We square the error:

$$
\left(\hat{y}^{(i)}-y^{(i)}\right)^2
$$

In our example:

$$
(-3)^2=9
$$

We square the error because otherwise positive and negative errors could cancel each other out.

For example:

$$
+3 + (-3) = 0
$$

This would make it look like there was no error, even though both predictions were wrong.

After squaring, both errors become positive:

$$
(+3)^2=9 \qquad (-3)^2=9
$$

---

## What does the Summation do?

The symbol $\sum$ means **sum**.

So we calculate the squared error for **every training example** and then add all of them together.

```text
Squared error 1
      +
Squared error 2
      +
Squared error 3
      +
Squared error 4
      +
Squared error 5
      ↓
Total squared error
```

Mathematically, this is represented by:

$$
\sum_{i=1}^{m}\left(\hat{y}^{(i)}-y^{(i)}\right)^2
$$

The $i=1$ means we start from the **first training example**, and $m$ means we continue until the **last training example**.

Then we divide by $2m$:

$$
\frac{1}{2m}
$$

where $m$ is the number of training examples.

This gives us the **cost value** $J(w,b)$ for our model:

$$
J(w,b)=\frac{1}{2m}\sum_{i=1}^{m}\left(\hat{y}^{(i)}-y^{(i)}\right)^2
$$

---

## What Does the Cost Tell Us?

The cost gives us a measure of how well our model is predicting.

### Lower Cost

A **lower cost** generally means the predictions are closer to the actual values.

### Higher Cost

A **higher cost** generally means the predictions are farther from the actual values.

So we can think of it like:

```text
Lower Cost  → Better predictions
Higher Cost → Worse predictions
```

The goal of linear regression is therefore to find values of $w$ and $b$ that produce a **low cost**.

In the next step, we will implement this cost function with our dataset and see exactly how the calculation works.

````python
import numpy as np

def compute_cost(y, f_wb):
    """
    Compute the cost function for linear regression.

    Parameters
    ----------
    y : ndarray (m,)
        Actual target values for m training examples.
    f_wb : ndarray (m,)
        Predicted values (model output) for the same m training examples.

    Returns
    -------
    cost : float
        The computed cost J(w, b).
    """
    m = len(y)  # number of training examples

    squared_errors = (f_wb - y) ** 2
    sum_squared_errors = np.sum(squared_errors)
    cost = (1 / (2 * m)) * sum_squared_errors

    return cost
````

````python
#Here we compute the cost using the actual values y and the predicted values f_wb
cost = compute_cost(y, f_wb)

print("Cost:", cost)
````

````python

# Here is the graph of the cost function with respect to w and b. The cost function is a convex function, which means that it has a single global minimum. The minimum point corresponds to the optimal values of w and b that minimize the cost function
plt.figure(figsize=(8, 6))
plt.plot(w,cost, 'ro', label='Current Cost')
plt.xlabel('Weight (w)')
plt.ylabel('Cost (J(w, b))')
plt.title('Cost Function vs Weight')
plt.legend()
plt.grid()
plt.show()
````

# Gradient and Gradient Descent

## What is a Gradient?

The **gradient** tells us the direction and rate at which the cost function $J(w,b)$ is changing with respect to each parameter ($w$ and $b$). In other words, it tells us **which way to adjust $w$ and $b$** to reduce the cost.

We use this gradient inside an update rule to move $w$ and $b$ toward values that minimize the cost.

## The Update Rule

$$
w = w - \alpha \frac{\partial J(w,b)}{\partial w}
$$

$$
b = b - \alpha \frac{\partial J(w,b)}{\partial b}
$$

Where:

* $\alpha$ (alpha) = the **learning rate** — controls how big a step we take on each update
* $\frac{\partial J(w,b)}{\partial w}$ = the gradient (slope) of the cost function with respect to $w$
* $\frac{\partial J(w,b)}{\partial b}$ = the gradient (slope) of the cost function with respect to $b$

These two equations give us the **new, updated values** of $w$ and $b$ on each step.

## The Gradient Formulas

To actually compute these gradients, we use:

$$
\frac{\partial J(w,b)}{\partial w} = \frac{1}{m}\sum_{i=1}^{m}\left(\hat{y}^{(i)}-y^{(i)}\right)x^{(i)}
$$

$$
\frac{\partial J(w,b)}{\partial b} = \frac{1}{m}\sum_{i=1}^{m}\left(\hat{y}^{(i)}-y^{(i)}\right)
$$

In plain words:

* For $\frac{\partial J}{\partial w}$: take the error ($\hat{y}^{(i)} - y^{(i)}$) for each training example, multiply it by that example's $x^{(i)}$, then average over all $m$ examples.
* For $\frac{\partial J}{\partial b}$: take the same error for each example, and simply average it over all $m$ examples (no multiplication by $x$).

## What is Gradient Descent?

**Gradient Descent** is the algorithm that repeatedly applies the update rule above, over many iterations, to gradually reduce the cost $J(w,b)$ until it reaches (or gets close to) its minimum value.

Each iteration:
1. Compute the gradients $\frac{\partial J}{\partial w}$ and $\frac{\partial J}{\partial b}$ using the current $w$ and $b$.
2. Update $w$ and $b$ **simultaneously** using the update rule.
3. Recompute the cost with the new $w$ and $b$.
4. Repeat until the cost stops decreasing significantly (convergence) or a set number of iterations is reached.

This is how the model's error gradually shrinks over time.

---

## Summary: The Full Flow

```text
Prediction  →  Cost  →  Gradient  →  Update (w, b)  →  Check Cost
     ↑                                                      |
     └──────────────────── repeat ─────────────────────────┘
```

1. **Prediction** — compute $\hat{y} = wx + b$ using current $w, b$
2. **Cost** — measure how far off the predictions are using $J(w,b)$
3. **Gradient** — compute the direction/rate of change of the cost with respect to $w$ and $b$
4. **Update** — adjust $w$ and $b$ using the gradient and learning rate $\alpha$
5. **Check Cost** — recompute the cost with the new $w, b$

If the cost is **low**, the model's predictions are close to the actual values → the model is accurate.
If the cost is still **high**, the loop continues, gradually improving $w$ and $b$ until the cost is minimized.

````python
# Gradient Descent needs the gradient to know
# how the cost function changes with respect to w and b.

def compute_gradient(x, y, f_wb):
    """
    Compute the gradient of the cost function
    with respect to w and b.

    x    : input values
    y    : actual target values
    f_wb : model predictions
    """

    # Number of training examples
    m = len(x)

    # Calculate the error for each training example
    # Error = actual value - predicted value
    error = f_wb - y

    # Calculate the contribution of each example
    # to the gradient with respect to w.
    error_x = error * x

    # Average all the values to get the gradient
    # with respect to w.
    dj_dw = (1 / m) * np.sum(error_x)

    # Average all the values to get the gradient
    # with respect to b.
    dj_db = (1 / m) * np.sum(error)

    # Return both gradient values
    return dj_dw, dj_db
````

# Gradient vs Gradient Descent

A common confusion in Machine Learning is:

> **What is the difference between Gradient and Gradient Descent?**

The easiest way to understand them is to imagine that we are standing on a hill.

Our goal is to reach the **bottom of the hill**.

In Machine Learning:

- The **hill/landscape** represents the Cost Function.
- Our position is determined by `w` and `b`.
- The **bottom** represents the minimum cost.

## 🏔️ Imagine You Are Standing on a Hill

Imagine you are standing somewhere on a hill with your eyes closed.

You want to reach the bottom.

But you cannot see where the bottom is.

So you need to know:

> **"Which direction is uphill?"**

The answer to this question is given by the **gradient**.

The gradient tells us the direction in which the cost is increasing.

But we don't want to go uphill.

We want to go downhill.

Therefore, we move in the **opposite direction of the gradient**.


## 1. What is the Gradient?

The **gradient is information about the slope of the Cost Function**.

It tells us:

1. Which direction makes the cost increase.
2. How strongly the cost changes in that direction.

For linear regression, we have two parameters:

- `w` → weight
- `b` → bias

Therefore, we calculate two derivatives:

`dw = ∂J/∂w`

`db = ∂J/∂b`

Together:

`Gradient = [dw, db]`

Think of the gradient as someone telling you:

> **"The cost goes uphill in this direction."**

### Important:

The gradient itself does **NOT** update `w` and `b`.

It only gives us information about how the cost is changing.

````python
compute_gradient(x, y, f_wb)
# The output of the compute_gradient function 
````

## 2. What is Gradient Descent?

Now we know which direction is uphill.

But our goal is to reach the bottom.

So we move in the **opposite direction of the gradient**.

This process is called:

# Gradient Descent

Gradient Descent is an **optimization algorithm**.

It repeatedly:

1. Calculates the gradient.
2. Looks at the direction of increasing cost.
3. Moves in the opposite direction.
4. Updates `w` and `b`.
5. Calculates the cost again.
6. Repeats the process.

The goal is to gradually reduce the Cost Function.

````python
def gradient_descent(x, y, f_wb, alpha, compute_gradient, w, b):

    # Calculate the gradient of the cost function.
    # dj_dw tells us how the cost changes with respect to w.
    # dj_db tells us how the cost changes with respect to b.
    dj_dw, dj_db = compute_gradient(x, y, f_wb)

    # Update w using Gradient Descent.
    # We move w in the opposite direction of the gradient.
    # alpha (learning rate) controls how big the step is.
    w = w - alpha * dj_dw

    # Update b using Gradient Descent.
    # Again, we move in the opposite direction of the gradient.
    b = b - alpha * dj_db

    # Return the new/updated values of w and b.
    return w, b
````

````python
gradient_descent(x, y, f_wb, 0.01, compute_gradient,w,b)
# This function performs one step of gradient descent, updating the values of w and b based on the computed gradients and the learning rate alpha.
````

## Before Training

So far, we have chosen `w` and `b` ourselves.

For example:

```text
w = 6
b = 40
```

Using these values, our model creates a prediction line.

However, the line does not fit the actual data very well.

We can also calculate the cost to measure how far our predictions are from the actual values.

Now the question is:

> **Can we automatically find better values for `w` and `b`?**

This is where **Gradient Descent** comes in.

Instead of manually changing `w` and `b`, gradient descent will repeatedly update them to reduce the cost.


````python
w = 6
b = 40

alpha = 0.01
iterations = 1000

for i in range(iterations):

    # Calculate predictions
    f_wb = w * x + b

    # Calculate gradient
    dj_dw, dj_db = compute_gradient(x, y, f_wb)

    # Update w and b
    w = w - alpha * dj_dw
    b = b - alpha * dj_db

print("Final w:", w)
print("Final b:", b)
````

## Visualizing the Trained Model

We have now trained our linear regression model using Gradient Descent.

The model has learned values for `w` and `b`.

Now let's use these learned values to make predictions and plot the final prediction line together with our actual data.

This allows us to visually see how well the trained model fits the data.


````python
# Make predictions using the trained w and b
final_predictions = w * x + b

# Plot actual data
plt.figure(figsize=(8, 5))

plt.scatter(
    x,
    y,
    label="Actual Data"
)

# Plot trained prediction line
plt.plot(
    x,
    final_predictions,
    label=f"Trained Model (w={w:.2f}, b={b:.2f})"
)

plt.xlabel("Hours Studied")
plt.ylabel("Exam Marks")
plt.title("Linear Regression After Training")

plt.legend()
plt.grid()

plt.show()
````
